```python
from pathlib import Path
from zipfile import ZipFile, ZIP_DEFLATED
```

```python
base_dir = Path("/Users/htelg/nfs/grad/campaign/frc/2025/AOD").resolve()

folders_to_zip = [
    Path("/Users/htelg/nfs/grad/campaign/frc/2025/AOD/US_NOAA_MFR_8/netcdf/v1.0/"),
    Path("/Users/htelg/nfs/grad/campaign/frc/2025/AOD/US_NOAA_MFR_8/aeronet_format/v1.0/"),
    Path("/Users/htelg/nfs/grad/campaign/frc/2025/AOD/US_NOAA_MFR_9/netcdf/v1.0/"),
    Path("/Users/htelg/nfs/grad/campaign/frc/2025/AOD/US_NOAA_MFR_9/aeronet_format/v1.0/"),
    
]

out_zip = Path("/Users/htelg/nfs/grad/campaign/frc/2025/AOD/US_NOAA_MFR.zip")

with ZipFile(out_zip, mode="w", compression=ZIP_DEFLATED) as zf:
    for folder in folders_to_zip:
        folder = folder.resolve()

        for file in folder.rglob("*"):
            if file.is_file():
                arcname = file.relative_to(base_dir)
                zf.write(file, arcname=arcname)

print(f"Wrote {out_zip}")
```
